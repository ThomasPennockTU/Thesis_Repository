```python
# -*- coding: utf-8 -*-
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import pandas as pd
import numpy as np
from matplotlib.colors import LinearSegmentedColormap

# === CONFIGURATION ===
csv_path = 'uplift_downdrift_lastframe_maxcorners_05-1950_30.csv'

# Load CSV
df = pd.read_csv(csv_path)

# Keep only the last row
row = df.iloc[-1]

# Extract time
time_value = row["Time"]
Hs = 0.6  # or load from file if you saved it

# Identify uplift columns
uplift_cols = [c for c in df.columns if c.startswith("uplift_")]
block_names = [c.replace("uplift_", "") for c in uplift_cols]

# Make a Series mapping block name → uplift value
uplift_data = pd.Series(
    row[uplift_cols].values,
    index=block_names
)

# === NEW: find and print highest uplift ===
uplift_data_clean = uplift_data.dropna()
if not uplift_data_clean.empty:
    max_block = uplift_data_clean.idxmax()
    max_value = uplift_data_clean.max()
    print("✅ Highest uplift value:")
    print("   Block: %s" % max_block)
    print("   Value: %.6f m" % max_value)
else:
    print("ℹ️ No valid uplift data found.")

# Plotting parameters
full_w = 0.5
full_h = 0.5
half_w = 0.225
half_h = 0.5
horizontal_spacing = 0.05
v_spacing = 0.0
rows = 22

grid_width = (9 * full_w) + (8 * horizontal_spacing)
grid_height = rows * full_h

# Normalize color
norm = plt.Normalize(vmin=0, vmax=0.150)

# Colormap
colors = ["#10b151", "#f6eb01", "#ffc112", "#ed2024"]
cmap = LinearSegmentedColormap.from_list("uplift_cmap", colors, N=256)
no_data_color = "lightgray"

# Plot grid
fig, ax_grid = plt.subplots(figsize=(10, 12))

for row_num in range(1, rows + 1):
    current_y = (rows - row_num) * (full_h + v_spacing)
    is_half = (row_num % 2 == 1)
    
    if is_half:
        block_sequence = (
            [(half_w, f"B-{row_num}-1")] +
            [(full_w, f"B-{row_num}-{col}") for col in range(2, 10)] +
            [(half_w, f"B-{row_num}-10")]
        )
    else:
        block_sequence = [
            (full_w, f"B-{row_num}-{col}") for col in range(1, 10)
        ]

    x_pos = 0.0
    for block_w, block_name in block_sequence:
        if block_name in uplift_data:
            value = uplift_data[block_name]
            if pd.isnull(value):
                color = no_data_color
            else:
                color = cmap(norm(value))
        else:
            color = no_data_color

        rect = patches.Rectangle(
            (x_pos, current_y),
            block_w,
            full_h,
            edgecolor='black',
            facecolor=color,
            linewidth=1
        )
        ax_grid.add_patch(rect)
        x_pos += block_w + horizontal_spacing

ax_grid.set_aspect('equal')
ax_grid.set_xlim(0, grid_width)
ax_grid.set_ylim(0, grid_height)
ax_grid.axis('off')

# Colorbar
sm = plt.cm.ScalarMappable(cmap=cmap, norm=norm)
sm.set_array([])
cbar = plt.colorbar(sm, ax=ax_grid, orientation='vertical')
cbar.set_label('Relative Uplift [m]')

# Title
ax_grid.set_title(
    f"Block Grid Uplift - Hs={Hs:.2f} m   Time={time_value:.2f} s",
    fontsize=16,
    fontweight='bold'
)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
import copy
from matplotlib.colors import ListedColormap, BoundaryNorm

# Set figure and axes background colors to white
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

# Set STIX font and math support, with fontsize 16
plt.rcParams["font.family"] = "STIX Two Text"
plt.rcParams["mathtext.fontset"] = "stix"
plt.rcParams["font.size"] = 16

# Young's modulus values (rows)
youngs_modulus_values = [2, 3, 7.5, 30]  # in GPa

# Density values (columns)
density_values = [2300, 2100, 2000, 1950]  # in kg/m³

# First matrix
results_matrix1 = np.array([
    [0.01, 0.0, 0.0, 1],
    [0.01, 0.0, 1, 0.0],
    [0.41, 0.71, 0.0, 0.0],
    [0.01, 0.41, 0.01, 1]
])

# Second matrix
results_matrix2 = np.array([
    [0.01, 0.0, 0.0, 0.01],
    [0.01, 0.0, 0.01, 0.0],
    [0.01, 0.01, 0.0, 0.0],
    [0.01, 0.41, 0.41, 0.71]
])

# Define colors and boundaries

colors = ["#10b151", "#f6eb01", "#ffc112", "#ed2024"]
boundaries = [0.0, 0.2, 0.5, 0.8, 1.0]

# Create discrete colormap and norm
cmap = ListedColormap(colors)
norm = BoundaryNorm(boundaries, ncolors=cmap.N)

# Function to plot one matrix on a given axis
def plot_matrix(ax, matrix, density_values, youngs_modulus_values, cmap, norm, opacity):
    # Mask zero values
    masked_matrix = np.ma.masked_where(matrix == 0, matrix)

    # Create grid edges
    x_edges = np.arange(len(density_values) + 1)
    y_edges = np.arange(len(youngs_modulus_values) + 1)

    # Plot with pcolormesh
    # c = ax.pcolormesh(
    #     x_edges,
    #     y_edges,
    #     masked_matrix,
    #     cmap=cmap,
    #     norm=norm,
    #     edgecolors='white',
    #     linewidth=8,
    #     alpha=opacity
    # )

    c = ax.pcolormesh(
        x_edges,
        y_edges,
        masked_matrix,
        cmap=cmap,
        norm=norm,
        edgecolors='black',
        linewidth=0.5,
        alpha=opacity
    )

    # Center ticks on squares
    ax.set_xticks(np.arange(len(density_values)) + 0.5)
    ax.set_yticks(np.arange(len(youngs_modulus_values)) + 0.5)

    # Set tick labels
    ax.set_xticklabels(density_values)
    ax.set_yticklabels(youngs_modulus_values)

    # Move x-axis to top
    ax.xaxis.tick_top()
    ax.xaxis.set_label_position('top')

    # Instead of set_xlabel, plot text higher above the axis
    ax.text(
        0.5, 1.1,
        'Density (kg/m³)',
        transform=ax.transAxes,
        ha='center',
        va='bottom'
    )

    # Set y-axis label as usual
    ax.set_ylabel("Young's Modulus (GPa)")

    # Force squares
    ax.set_aspect('equal')

    # Remove spines
    for spine in ax.spines.values():
        spine.set_visible(False)

    return c

# Opacity
opacity = 0.7

# Create figure and two subplots side by side
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Plot first matrix
c1 = plot_matrix(
    axes[0],
    results_matrix1,
    density_values,
    youngs_modulus_values,
    cmap,
    norm,
    opacity
)



# Plot second matrix
c2 = plot_matrix(
    axes[1],
    results_matrix2,
    density_values,
    youngs_modulus_values,
    cmap,
    norm,
    opacity
)

# Add colorbar
fig.subplots_adjust(right=0.85)
cbar_ax = fig.add_axes([0.99, 0.15, 0.03, 0.7])

# Create colorbar with boundaries
cbar = fig.colorbar(
    c1,
    cax=cbar_ax,
    boundaries=boundaries,
    ticks=[0.1, 0.35, 0.65, 0.9],
)
cbar.ax.set_yticklabels(['1', '2', '3', '4'])
# cbar.set_label('Test Result Value')

# Control border around the colorbar
for spine in cbar.ax.spines.values():
    spine.set_linewidth(0.5)          # adjust thickness here
    spine.set_edgecolor("black")      # adjust color here

# Add figure-level text in the top-left corner
fig.text(
    0.01, 0.95,
    "Hs=0.6",
    ha='left',
    va='top',
    fontsize=16,
    fontweight='bold',
    color='black'
)

# Add figure-level text in the top-middle
fig.text(
    0.51, 0.95,
    "Hs=0.5",
    ha='left',
    va='top',
    fontsize=16,
    fontweight='bold',
    color='black'
)

plt.tight_layout()
plt.savefig('results_matrices.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap, BoundaryNorm

# White background
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'

# STIX font + size
plt.rcParams["font.family"] = "STIX Two Text"
plt.rcParams["mathtext.fontset"] = "stix"
plt.rcParams["font.size"] = 16

# Axes values
youngs_modulus_values = [2, 3, 7.5, 30]       # rows (GPa)
density_values = [2300, 2100, 2000, 1950]     # cols (kg/m³)

# Example matrices
results_matrix1 = np.array([
    [0.01, 0.0, 0.0, 1],
    [0.01, 0.0, 1, 0.0],
    [0.41, 0.71, 0.0, 0.0],
    [0.01, 0.41, 0.01, 1]
])

results_matrix2 = np.array([
    [0.01, 0.0, 0.0, 0.01],
    [0.01, 0.0, 0.01, 0.0],
    [0.01, 0.01, 0.0, 0.0],
    [0.01, 0.41, 0.41, 0.71]
])

# Discrete colormap
colors = ["#10b151", "#f6eb01", "#ffc112", "#ed2024"]
boundaries = [0.0, 0.2, 0.5, 0.8, 1.0]
cmap = ListedColormap(colors)
norm = BoundaryNorm(boundaries, ncolors=cmap.N)

def plot_matrix(ax, matrix, density_values, youngs_modulus_values, cmap, norm, opacity, text_matrix=None):
    masked_matrix = np.ma.masked_where(matrix == 0, matrix)

    x_edges = np.arange(len(density_values) + 1)
    y_edges = np.arange(len(youngs_modulus_values) + 1)

    c = ax.pcolormesh(
        x_edges, y_edges, masked_matrix,
        cmap=cmap, norm=norm,
        edgecolors='black', linewidth=0.5, alpha=opacity
    )

    ax.set_xticks(np.arange(len(density_values)) + 0.5)
    ax.set_yticks(np.arange(len(youngs_modulus_values)) + 0.5)

    ax.set_xticklabels(density_values)
    ax.set_yticklabels(youngs_modulus_values)

    ax.xaxis.tick_top()
    ax.xaxis.set_label_position('top')
    ax.text(0.5, 1.1, 'Density (kg/m³)', transform=ax.transAxes, ha='center', va='bottom')
    ax.set_ylabel("Young's Modulus (GPa)")

    ax.set_aspect('equal')
    for spine in ax.spines.values():
        spine.set_visible(False)

    if text_matrix is not None:
        for i in range(matrix.shape[0]):
            for j in range(matrix.shape[1]):
                txt = text_matrix[i, j]
                if txt is not None:
                    ax.text(j + 0.5, i + 0.5, txt, ha='center', va='center', fontsize=14, color='black')

    return c

opacity = 0.7

# Labels for diagonals
# labels_diag_main = ["C30", r"DS$_1$", r"DS$_2$", r"DS$_3$"]
labels_diag_other = [r"DS$_3$", r"DS$_2$", r"DS$_1$", "C30"]


# Build text matrices with both diagonals
text_matrix1 = np.full(results_matrix1.shape, None, dtype=object)
text_matrix2 = np.full(results_matrix2.shape, None, dtype=object)

n = results_matrix1.shape[0]
for idx in range(n):
    # text_matrix1[idx, idx] = labels_diag_main[idx]
    text_matrix1[idx, n - idx - 1] = labels_diag_other[idx]

    # text_matrix2[idx, idx] = labels_diag_main[idx]
    text_matrix2[idx, n - idx - 1] = labels_diag_other[idx]

# Figure
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

c1 = plot_matrix(axes[0], results_matrix1, density_values, youngs_modulus_values, cmap, norm, opacity, text_matrix=text_matrix1)
c2 = plot_matrix(axes[1], results_matrix2, density_values, youngs_modulus_values, cmap, norm, opacity, text_matrix=text_matrix2)

# Colorbar
fig.subplots_adjust(right=0.85)
cbar_ax = fig.add_axes([0.99, 0.15, 0.03, 0.7])
cbar = fig.colorbar(c1, cax=cbar_ax, boundaries=boundaries, ticks=[0.1, 0.35, 0.65, 0.9])
cbar.ax.set_yticklabels(['1', '2', '3', '4'])

for spine in cbar.ax.spines.values():
    spine.set_linewidth(0.5)
    spine.set_edgecolor("black")

# Headers
fig.text(0.01, 0.95, "Hs=0.6", ha='left', va='top', fontsize=16, fontweight='bold', color='black')
fig.text(0.51, 0.95, "Hs=0.5", ha='left', va='top', fontsize=16, fontweight='bold', color='black')

plt.tight_layout()
plt.savefig('results_matrices_with_two_diagonals.pdf', dpi=300, bbox_inches='tight')
plt.show()
```
