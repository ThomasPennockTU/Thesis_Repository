```python
import numpy as np
import pandas as pd
from scipy import stats
import statsmodels.api as sm
from statsmodels.formula.api import ols
from statsmodels.stats.multicomp import pairwise_tukeyhsd

# ---------------------------
# Max Pressure Mixes
# ---------------------------

# 5 test results for each mix (replace these with your real measurements)
mix_0 = [33.23, 31.13, 35.26, 34.20, 32.09]
mix_20 = [11.34, 12.40, 11.32, 12.28, 11.35]
mix_40 = [5.27, 5.70, 5.86, 5.99, 5.25]
mix_60 = [5.17, 4.18, 4.24, 4.65, 4.89]

# ---------------------------
# NO NEED TO EDIT BELOW HERE
# ---------------------------

# Combine into single dataframe
mix_values = mix_0 + mix_20 + mix_40 + mix_60
mix_labels = (
    [0]*5 +
    [20]*5 +
    [40]*5 +
    [60]*5
)

df = pd.DataFrame({
    'Mix': mix_labels,
    'Strength': mix_values
})

# Display your input data
print("\nYour input data:")
print(df.groupby('Mix')['Strength'].describe())

# Run one-way ANOVA
model = ols('Strength ~ C(Mix)', data=df).fit()
anova_table = sm.stats.anova_lm(model, typ=2)

print("\nANOVA table:")
print(anova_table)

# Check p-value
if anova_table['PR(>F)'][0] < 0.05:
    tukey = pairwise_tukeyhsd(
        endog=df['Strength'],
        groups=df['Mix'],
        alpha=0.05
    )
    print("\nTukey HSD results:")
    print(tukey.summary())
else:
    print("\nNo significant differences found between mixes.")
```
